```python
Initial_Load = ""
Prev_Run_Dt = "12/12/2012 12:12:12"
Prev_Run_Dt = ""

query_1 = f"""SELECT
  DEPT_ID AS DEPT_ID,
  DEPT_CODE AS DEPT_CODE,
  DESCRIPTION AS DESCRIPTION,
  CREATE_DATE_TIME AS CREATE_DATE_TIME,
  MOD_DATE_TIME AS MOD_DATE_TIME,
  USER_ID AS USER_ID,
  WHSE AS WHSE,
  MISC_TXT_1 AS MISC_TXT_1,
  MISC_TXT_2 AS MISC_TXT_2,
  MISC_NUM_1 AS MISC_NUM_1,
  MISC_NUM_2 AS MISC_NUM_2,
  PERF_GOAL AS PERF_GOAL,
  VERSION_ID AS VERSION_ID,
  CREATED_DTTM AS CREATED_DTTM,
  LAST_UPDATED_DTTM AS LAST_UPDATED_DTTM,
  ROW_NUMBER() OVER (ORDER BY DEPT_ID) AS Monotonically_Increasing_Id
FROM
  Shortcut_to_NUODATA_TEST_DEPARTMENT_0
WHERE
  {Initial_Load} (
    DATE_TRUNC('DAY', CREATE_DATE_TIME) >= DATEADD('DAY', -1, DATE_TRUNC('DAY', TO_TIMESTAMP_NTZ('{Prev_Run_Dt}', 'MM/DD/YYYY HH24:MI:SS')))
    OR DATE_TRUNC('DAY', MOD_DATE_TIME) >= DATEADD('DAY', -1, DATE_TRUNC('DAY', TO_TIMESTAMP_NTZ('{Prev_Run_Dt}', 'MM/DD/YYYY HH24:MI:SS')))
    OR DATE_TRUNC('DAY', CREATED_DTTM) >= DATEADD('DAY', -1, DATE_TRUNC('DAY', TO_TIMESTAMP_NTZ('{Prev_Run_Dt}', 'MM/DD/YYYY HH24:MI:SS')))
    OR DATE_TRUNC('DAY', LAST_UPDATED_DTTM) >= DATEADD('DAY', -1, DATE_TRUNC('DAY', TO_TIMESTAMP_NTZ('{Prev_Run_Dt}', 'MM/DD/YYYY HH24:MI:SS')))
  )
  AND 1 = 1"""

df_1 = session.sql(query_1)
df_1.collect()
df_1.create_or_replace_temp_view("SQ_Shortcut_to_TEST_DEPARTMENT_1")
```
